```python
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder

cat_cols = [
    'Gender',
    'City_Type',
    'Current_Car_Type',
    'Home_Charging_Possible',
    'Subsidy_Available',
    'Range_Anxiety_Level'
]

encoder = OrdinalEncoder()

df_train[cat_cols] = encoder.fit_transform(df_train[cat_cols])
df_test[cat_cols] = encoder.transform(df_test[cat_cols])

encoder_y = LabelEncoder()

df_train['Will_Buy_EV'] = encoder_y.fit_transform(df_train['Will_Buy_EV'])
```

```python
df_test.head()
```

```python
df_train.head()
```

```python
df_train=df_train.drop(columns=['id'])
```

```python
df_test = df_test.drop(columns=['id'],axis=0)
```

```python
df_test.columns
```

```python
X_train = df_train.drop(columns=['Will_Buy_EV'])
y_train = df_train['Will_Buy_EV']

X_test = df_test
```

```python
print(df_train.shape)
print(X_train.shape)
print(y_train.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)
```

```python
print(df_train.shape)
print(X_train.shape)
print(y_train.shape)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_train)

accuracy_score(y_train, y_pred)
```

```python
y_pred = model.predict_proba(X_test)[:, 1]

submission = pd.DataFrame({
    'id': save['id'],
    'Will_Buy_EV': y_pred
})

submission.to_csv('submission.csv', index=False)
```
