```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

ordinal_cols = [
    'Gender',
    'Home_Charging_Possible',
    'Subsidy_Available',
    'Range_Anxiety_Level'
]

onehot_cols = [
    'City_Type',
    'Current_Car_Type'
]

encoder = OrdinalEncoder()

df_train[ordinal_cols] = encoder.fit_transform(
    df_train[ordinal_cols]
)

df_test[ordinal_cols] = encoder.transform(
    df_test[ordinal_cols]
)


encoder_ohe = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

train_encoded = encoder_ohe.fit_transform(
    df_train[onehot_cols]
)

test_encoded = encoder_ohe.transform(
    df_test[onehot_cols]
)


train_encoded = pd.DataFrame(
    train_encoded,
    columns=encoder_ohe.get_feature_names_out(onehot_cols),
    index=df_train.index
)

test_encoded = pd.DataFrame(
    test_encoded,
    columns=encoder_ohe.get_feature_names_out(onehot_cols),
    index=df_test.index
)


df_train = df_train.drop(
    onehot_cols,
    axis=1
)

df_test = df_test.drop(
    onehot_cols,
    axis=1
)


df_train = pd.concat(
    [df_train, train_encoded],
    axis=1
)

df_test = pd.concat(
    [df_test, test_encoded],
    axis=1
)
```

```python
df_test.head()
```

```python
df_train.head()
```

```python
df_train = df_train.drop(
    columns=['id']
)

df_test = df_test.drop(
    columns=['id']
)
```

```python
X = df_train.drop(
    columns=['Will_Buy_EV']
)

y = df_train['Will_Buy_EV']

X_test = df_test

print("X shape:", X.shape)
print("y shape:", y.shape)
print("X_test shape:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(
    max_iter=5000,
    random_state=42
)

lr_model.fit(
    X_scaled,
    y
)
```

```python
from sklearn.metrics import accuracy_score

y_pred_lr = lr_model.predict(
    X_scaled
)

accuracy_lr = accuracy_score(
    y,
    y_pred_lr
)

print("Logistic Regression Accuracy:", accuracy_lr)
```

```python
y_pred_lr_test = lr_model.predict_proba(
    X_test_scaled
)[:, 1]

submission_lr = pd.DataFrame({
    'id': save['id'],
    'Will_Buy_EV': y_pred_lr_test
})

submission_lr.to_csv(
    'submission.csv',
    index=False
)
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=1200,
    learning_rate=0.02,
    max_depth=4,
    min_child_weight=2,
    subsample=0.9,
    colsample_bytree=0.9,
    gamma=0.05,
    reg_alpha=0.05,
    reg_lambda=2,
    objective='binary:logistic',
    eval_metric='logloss',
    random_state=42
)

xgb_model.fit(
    X,
    y
)
```

```python
y_pred_xgb = xgb_model.predict(
    X
)

accuracy_xgb = accuracy_score(
    y,
    y_pred_xgb
)

print("XGBoost Accuracy:", accuracy_xgb)
```

```python
y_pred_xgb_test = xgb_model.predict_proba(
    X_test
)[:, 1]

submission_xgb = pd.DataFrame({
    'id': save['id'],
    'Will_Buy_EV': y_pred_xgb_test
})

submission_xgb.to_csv(
    'submission.csv',
    index=False
)

print("XGBoost submission created")
```
